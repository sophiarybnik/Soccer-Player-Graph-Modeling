```python
from pathlib import Path
import sys
from src.data.loaders import load_matches
from src.graphs.visualize import plot_event_graph
import torch
```

```python
PROJECT_ROOT = Path.cwd().parent
sys.path.insert(0, str(PROJECT_ROOT))

graphs_path = PROJECT_ROOT / "data" / "graphs" / "train.pt"
graphs = torch.load(graphs_path)

matches_path = PROJECT_ROOT / "statsbomb_data"
matches_df = load_matches(matches_path)
```

```python
for i in range(4):
    event_graph = graphs[i]
    competition_name = matches_df.iloc[i].competition.get('competition_name')
    season_name = matches_df.iloc[i].season.get('season_name')
    match_num = matches_df.iloc[i].match_id
    event_type = event_graph.metadata["event_type"]
    player_name = event_graph.metadata["player_name"]
    title = f"{event_type}: {player_name}\n {competition_name} | {season_name} | Match {match_num}"
    plot_event_graph(event_graph, title)
```
